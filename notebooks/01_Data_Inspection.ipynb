```python
# 1. Inspect missing children records
print("Missing Children Records:")
display(df_work[df_work["children"].isnull()])


# 2. Inspect negative ADR records
print("\nNegative ADR Records:")
display(df_work[df_work["adr"] < 0])


# 3. Inspect unusually high ADR records
print("\nADR Above 1000:")
display(df_work[df_work["adr"] > 1000])


# 4. Inspect bookings with no guests
print("\nBookings With No Guests:")
display(no_guest_bookings)


# 5. Check whether duplicate removal changes dataset size
print("\nDataset before duplicate removal:", df_work.shape)

print(
    "Dataset after duplicate removal:",
    df_work.drop_duplicates().shape
)
```

```python
# 1. Remove exact duplicate records
df_work = df_work.drop_duplicates()

print("Shape after removing duplicates:", df_work.shape)

```

```python
# 2. Handle missing values

# Children: only 4 missing values
df_work["children"] = df_work["children"].fillna(0)

# Country: retain records and mark missing country
df_work["country"] = df_work["country"].fillna("Unknown")

# Agent and company: missing means no recorded agent/company
df_work["agent"] = df_work["agent"].fillna(0)
df_work["company"] = df_work["company"].fillna(0)
```

```python
# 3. Handle unusual ADR values

# Remove the single negative ADR record
df_work = df_work[df_work["adr"] >= 0].copy()

# Keep the unusually high ADR record for now.
# We will assess its impact during EDA.


# 4. Investigate bookings with no guests
# Do not delete them automatically; retain for analysis.


# 5. Convert reservation date into datetime
df_work["reservation_status_date"] = pd.to_datetime(
    df_work["reservation_status_date"],
    errors="coerce"
)


# 6. Create useful analytical columns

# Total guests
df_work["total_guests"] = (
    df_work["adults"] +
    df_work["children"] +
    df_work["babies"]
)

# Total nights stayed
df_work["total_nights"] = (
    df_work["stays_in_weekend_nights"] +
    df_work["stays_in_week_nights"]
)

# Total booking revenue estimate
df_work["estimated_revenue"] = (
    df_work["adr"] * df_work["total_nights"]
)


# 7. Check the cleaned dataset

print("Final dataset shape:", df_work.shape)

print("\nRemaining missing values:")
print(df_work.isnull().sum().sort_values(ascending=False).head(10))

print("\nRemaining duplicates:", df_work.duplicated().sum())

print("\nData types:")
print(df_work.dtypes)


# 8. Save cleaned dataset

from pathlib import Path
output_path = Path("data/processed/hotel_bookings_cleaned.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df_work.to_csv(output_path, index=False)
print("\nCleaned dataset saved successfully!")
```
