```python
# 1. Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# 2. Load cleaned dataset

df = pd.read_csv("processed/hotel_bookings_cleaned.csv")

print("Cleaned dataset loaded successfully!")

# 3. Basic overview

print("\nDataset Shape:", df.shape)

print("\nFirst 5 Records:")
display(df.head())

print("\nDataset Information:")
df.info()

# 4. Statistical summary

display(df.describe())

# 5. Check missing values

print("\nRemaining Missing Values:")
print(df.isnull().sum().sort_values(ascending=False).head(10))

# 6. Check duplicate records

print("\nRemaining Duplicate Rows:", df.duplicated().sum())
```

```python
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Creating folder for charts 
os.makedirs("../visualizations", exist_ok=True)


# 1. GENERAL BOOKING ANALYSIS

print("TOTAL BOOKINGS:", len(df))
print("\nHotel-wise Booking Distribution:")
print(df["hotel"].value_counts())

print("\nCancellation Distribution:")
print(df["is_canceled"].value_counts())

print("\nOverall Cancellation Rate:",
      round(df["is_canceled"].mean() * 100, 2), "%")

print("\nAverage Lead Time:",
      round(df["lead_time"].mean(), 2), "days")

print("\nAverage Length of Stay:",
      round(df["total_nights"].mean(), 2), "nights")


# 2. HOTELWISE ANALYSIS

hotel_analysis = df.groupby("hotel").agg(
    total_bookings=("hotel", "count"),
    cancellation_rate=("is_canceled", "mean"),
    average_adr=("adr", "mean"),
    average_stay=("total_nights", "mean")
)

hotel_analysis["cancellation_rate"] *= 100

print("\nHOTELWISE ANALYSIS:")
display(hotel_analysis.round(2))


# 3. MONTHLY BOOKING ANALYSIS

month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

monthly_bookings = (
    df.groupby("arrival_date_month")
    .size()
    .reindex(month_order)
)

print("\nMONTHLY BOOKINGS:")
display(monthly_bookings)


# 4. MARKET SEGMENT ANALYSIS

market_analysis = df.groupby("market_segment").agg(
    bookings=("market_segment", "count"),
    cancellation_rate=("is_canceled", "mean")
)

market_analysis["cancellation_rate"] *= 100

print("\nMARKET SEGMENT ANALYSIS:")
display(market_analysis.round(2))


# 5. CUSTOMER TYPE ANALYSIS

print("\nCUSTOMER TYPE DISTRIBUTION:")
display(df["customer_type"].value_counts())


# 6. LENGTH OF STAY ANALYSIS

print("\nLENGTH OF STAY ANALYSIS:")
display(df["total_nights"].describe())


# 7. CORRELATION ANALYSIS

numeric_columns = [
    "is_canceled",
    "lead_time",
    "total_nights",
    "adults",
    "children",
    "previous_cancellations",
    "booking_changes",
    "adr",
    "total_of_special_requests",
    "total_guests"
]

correlation = df[numeric_columns].corr()

print("\nCORRELATION MATRIX:")
display(correlation.round(2))


# Chart 1: Bookings by Hotel

plt.figure()
sns.countplot(data=df, x="hotel")
plt.title("Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
plt.tight_layout()
plt.savefig("../visualizations/01_bookings_by_hotel.png")
plt.show()
```

```python
# Chart 2: Cancellation Rate by Hotel

plt.figure()
sns.barplot(data=df, x="hotel", y="is_canceled")
plt.title("Cancellation Rate by Hotel")
plt.xlabel("Hotel Type")
plt.ylabel("Cancellation Rate")
plt.tight_layout()
plt.savefig("../visualizations/02_cancellation_by_hotel.png")
plt.show()
```

```python
# Chart 3: Monthly Booking Trends

plt.figure()
monthly_bookings.plot(kind="bar")
plt.title("Monthly Booking Distribution")
plt.xlabel("Arrival Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../visualizations/03_monthly_bookings.png")
plt.show()
```

```python
# Chart 4: Bookings by Market Segment

plt.figure()
sns.countplot(
    data=df,
    y="market_segment",
    order=df["market_segment"].value_counts().index
)
plt.title("Bookings by Market Segment")
plt.xlabel("Number of Bookings")
plt.ylabel("Market Segment")
plt.tight_layout()
plt.savefig("../visualizations/04_market_segment.png")
plt.show()
```

```python

# Chart 5: Length of Stay Distribution

plt.figure()
sns.histplot(
    data=df[df["total_nights"] <= 15],
    x="total_nights",
    bins=15
)
plt.title("Distribution of Length of Stay")
plt.xlabel("Total Nights")
plt.ylabel("Number of Bookings")
plt.tight_layout()
plt.savefig("../visualizations/05_length_of_stay.png")
plt.show()
```

```python
# Chart 6: Average Daily Rate Distribution

plt.figure()
sns.histplot(
    data=df[df["adr"] <= 500],
    x="adr",
    bins=40
)
plt.title("Average Daily Rate Distribution (ADR up to 500)")
plt.xlabel("Average Daily Rate")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("../visualizations/06_adr_distribution.png")
plt.show()
```

```python
# Chart 7: Correlation Heatmap

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Between Numerical Variables")
plt.tight_layout()
plt.savefig("../visualizations/07_correlation_heatmap.png")
plt.show()


print("\nEDA part succeeded!")
```
